# Linear Regression Exercise


## Problem
A teacher wants to understand whether the number of hours a student studies affects their exam score.
You have collected the following data from 6 students:
X (Hours Studied): [1, 2, 3, 4, 5, 6]
Y (Exam Score): [52, 55, 61, 66, 70, 74]
Your task is to use Linear Regression to find the relationship between study hours and exam scores.


## Goal
The goal of this exercise is to:
1. Calculate the mean of X and Y.
2. Calculate the slope (b1).
3. Calculate the intercept (b0).
4. Build the linear regression equation.
5. Use the equation to predict the exam score for a student who studies for 7 hours.
6. Implement the calculations in Python using NumPy, without using Scikit-learn.


## Regression Formula
The linear regression equation is:
y = b0 + b1x
Where:
* # y = predicted exam score
* # x = number of hours studied
## b0 = intercept
## b1 = slope

## Calculate the numerator of the slope

```python
X = np.array([1, 2, 3, 4, 5, 6])
Y = np.array([52, 55, 61, 66, 70, 74])

xmean = np.mean(X)
ymean = np.mean(Y)

print("xmean =", xmean)
print("ymean =", ymean)

sumxy = 0

for x, y in zip(X, Y):
    value = (x - xmean) * (y - ymean)
    print(x, y, value)
    sumxy += value

print("sumxy =", sumxy)
```

## Calculate the denominator of the slope formula


```python
sumx = 0 
for x in X:
    sumx += (x - xmean)*(x - xmean)
print (sumx)
```

## Calculate Slope

```python
b1 = sumxy/sumx
print(b1)
```

## Calculate intercept

```python
b0 = 0 
b0 = ymean - (b1 * xmean)
print (b0)
```

## The Regression Formula We Derived


```python
y = b0 + (b1*x)
```
