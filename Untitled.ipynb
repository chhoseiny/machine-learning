# House Price Prediction with Multiple Linear Regression

## Goal

In this exercise, we will build a Multiple Linear Regression model to predict house prices based on several features.

The dataset contains information about:
- House size
- Number of bedrooms
- House age
- Distance to the city center
- House price

The goal is to understand how multiple variables can be used together to predict a target value.

## Dataset

The dataset contains 15 house records.

### Features

- `Size` → House size in square meters
- `Bedrooms` → Number of bedrooms
- `Age` → Age of the house in years
- `DistanceToCenter` → Distance from the city center in kilometers

### Target

- `Price` → House price

## Split Data into Training and Test Sets

```python
x = df[["Size" , "Bedrooms" , "Age" , "DistanceToCenter"]]
y = df["Price"]
```

```python
msk = np.random.rand(len(y))<0.8
trainx = x[msk]
trainy = y[msk]


testx = x[~msk]
testy = y[~msk]
```

## modeling

using sklearn for calculate linearregression


```python
from sklearn import linear_model
regr = linear_model.LinearRegression()
train_x = np.asanyarray(trainx[['Size']])
train_y = np.asanyarray(trainy)
regr.fit(train_x , train_y)

print('Coeffitients: ' , regr.coef_)
print('Intercept: ' , regr.intercept_)
```

```python
plt.scatter(trainx.Size , trainy , color = 'blue')

plt.plot(train_x , regr.coef_* train_x + regr.intercept_ , '-r')
plt.xlabel("size")
plt.ylabel("price")
```

## Checking Regrresion


```python
from sklearn.metrics import r2_score
test_x = np.asanyarray(testx[['Size']])
test_y = np.asanyarray(testy)

test_y_ = regr.predict(test_x)

print("R2-score : %.2f" % r2_score(test_y , test_y_))
```
